```python
from google.colab.patches import cv2_imshow # For displaying images in Colab
import cv2
import numpy as np

# Load the image
img = cv2.imread('sample.jpg')

# Check if image loading failed and create a dummy image if it did
if img is None:
    print("Warning: 'sample.jpg' not found. Creating a dummy black and white image for demonstration.")
    # Create a 400x600 image with some basic shapes and text
    img = np.zeros((400, 600, 3), dtype=np.uint8)
    cv2.circle(img, (200, 200), 50, (0, 255, 0), -1) # Green circle
    cv2.rectangle(img, (350, 100), (500, 300), (255, 0, 0), -1) # Blue rectangle
    cv2.putText(img, "Dummy Image", (20, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)


# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Gaussian blur
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Perform Canny edge detection
edges = cv2.Canny(blur, 50, 150)

# Perform Harris corner detection
corners = cv2.cornerHarris(np.float32(gray), 2, 3, 0.04)
img_corners = img.copy()
# Ensure corners array is not empty before indexing and has values
if corners.size > 0 and corners.max() > 0:
    img_corners[corners > 0.01 * corners.max()] = [0, 0, 255] # Mark corners in red

# Perform SimpleBlobDetector detection
detector = cv2.SimpleBlobDetector_create()
keypoints = detector.detect(gray)
img_blobs = cv2.drawKeypoints(img, keypoints, None, (0, 255, 0), cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS) # Draw blobs in green

# Display the images using cv2_imshow for Colab compatibility
print("Original Image:")
cv2_imshow(img)
print("Grayscale Image:")
cv2_imshow(gray)
print("Blurred Image:")
cv2_imshow(blur)
print("Canny Edge Detection:")
cv2_imshow(edges)
print("Harris Corner Detection:")
cv2_imshow(img_corners)
print("SimpleBlobDetector (Blobs):")
cv2_imshow(img_blobs)

# cv2.waitKey(0) and cv2.destroyAllWindows() are not needed with cv2_imshow
```
